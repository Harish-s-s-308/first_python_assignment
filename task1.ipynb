```python
# Python program to ckeck a number is using if-else and logical operator


A = int(input("Enter first age:"))
B = int(input("Enter second age"))
C = int(input("Enter third age"))

if A>=B and A>=C:
    print("A is Older:")
elif B>=A and B>=C:
    print("B is Older:")
else:
    print("C is Older:")
```
