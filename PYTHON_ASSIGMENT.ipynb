```python
# This program adds two numbers and print the result

x = 10
y = 20
print(x+y)
```

```python
# This program print a message with the user Name

NAME = "HARISH"
print("HELLO",NAME)
```

```python
# This program swaps the values of two varaibles without using a third varaiables

A = 5
B = 10
A = A + B
B = A - B
A = A - B

print("A=",A)
print("B=",B)
```

```python
# This program takes the user age as input and displays 

AGE = int(input())

print("Your are",AGE,"Years old")
```

```python
# This program checks whether a numbers is even or odd

num = int(input())
if num % 2 == 0 :
    print("Even")
else :
    print("Odd")
```

```python
# This program performs arithmetic operations on two numbers and prints the results 

A = 15
B = 4

print("ADD:",A + B)
print("SUB:",A -B)
print("MULTI:",A * B)
print("DIV:",A /B)
print("FLOOR DIV:",A % B)
print("MODULUS:",A % B)
print("EXPONENT:",A ** B)
```

```python
# This program checks whether a Num is POSITIVE,NEGATIVE, OR ZERO


NUM = int(input())
if NUM > 0:
    print("POSITIVE")
elif NUM < 0:
    print("NEGATIVE")
else:
    print("ZERO")
```
