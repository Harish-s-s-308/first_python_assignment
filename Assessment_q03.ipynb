## *Find the Length of a List Using len()*

```python
mylist = [10,20,30,40,50]

length = len(mylist)

print("Length of the list:",length)
```

## *Find Maximum and Minimum Values Using max() and min()*

```python
mylist = [5,25,15,40,10]

maximum = max(mylist)
minimum = min(mylist)

print("Maximum Values:",maximum)
print("Minimum Values:",minimum)
```

## *Calculate the Total  of All Numbers Using sum()*

```python
mylist = [13,8,92,46,72]

total = sum(mylist)

print("Total:",total)
```

## *Sort a List in Ascending and Descending Order Using  sort()*

```python
mylist = [13,8,92,46,72]

mylist.sort()
print("Ascending order:",mylist)

mylist.sort(reverse = True)
print("Descendin order:",mylist)
```

## *Create a Tuple and Print It*

```python
mytuple = (13,8,92,46,72)

print("Tuple:",mytuple)
```

## *Access First, Middle,and Last Elements of a Tuple Using Indexing*

```python
mytuple = (13,8,92,46,72)

print("First element:",mytuple[0])
print("Middle element:",mytuple[2])
print("Last element:",mytuple[4])
```

## *Slice a Tuple (First 3 Elements and Last 3 Elements)*

```python
mytuple = (2,4,6,8,7,14,24,18)

print("First 3 elements:",mytuple[:3])
print("Last 3 elements:",mytuple[-3:])
```

## *Remove an Elements from a Set Using discard()*

```python
myset = {10,15,20,30,35,40,50,55}

myset.discard(55)

print("Updated set:",myset)
```

## *Remove an Elements from a Set Using pop()*

```python
myset = {10,15,20,30,40,50,55}

removed = myset.pop()

print("Removed element:",removed)
print("Updated set:",myset)
```

## *Update the Values of an Existing Key- Value Pair Using a Dictionary*

```python
mydict = {
    "Name":"Avanith",
    "Age":"20",
    "City":"Tiruchengode"
}

mydict["Age"]=21

print(mydict)
```

## *Remove the Last Inserted Key-Value Pair Using popitem()*

```python
mydict = {
    "Name":"Avanith",
    "Age":"20",
    "City":"Tiruchengode",
    "Contact":"123456"
}

mydict.popitem()

print(mydict)
```

## *Remove All Key-Value Pairs Using clear()*

```python
mydict = {
    "Name":"Avanith",
    "Age":"20",
    "City":"Tiruchengode",
    "Contact":"123456"
}

mydict.clear()

print(mydict)
```

## *Get All Keys Using keys()*

```python
mydict = {
    "Name":"Avanith",
    "Age":"20",
    "City":"Tiruchengode",
    "Contact":"123456"
}

print("All keys:",mydict.keys())
```

## *Get All Key-Value Pairs Using items()*

```python
mydict = {
    "Name":"Avanith",
    "Age":"20",
    "City":"Tiruchengode",
    "Contact":"123456"
}

print("All key-value pairs:",mydict.items())
```
