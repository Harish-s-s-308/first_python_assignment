```python
# pattern 1

x=int(input("Enter no of rows :"))
y=int(input("Enter no of colums :"))

a=1
while a<=x :
    b=1
    while b<=a :
        print(b,end="")
        b +=1
    a +=1
    print() 
```

```python
# pattern 2

x=int(input("Enter no of rows:"))
symbol=input("Enter your fav symbol:")

a=1
while a<=x:
    b=1
    while b<=a:
        print(symbol,end='')
        b +=1
    print()
    a +=1
```

```python
# pattern 5

x=int(input("Enter no of rows:"))
y=int(input("Enter no of colums:"))

for a in range(x):
    for b in range(y):
        print(b,end="")
    print()
```
