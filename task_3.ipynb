## 1. Print number form 1 to 10 using a for loop

```python
for x in range(1,10+1):
    print(x,end=" ")
```

## 2.Print numbers from 10 to 1 using a while loop

```python
x = 10
while x>0:
    print(x,end=" ")
    x -= 1
```

## 3.Write a program to print the multiplication table of a give number.

```python
num = int(input("Enter the number:"))
for x in range(1,10+1):
    y = num*x
    print(x,"*",num,"=",y)
```

## 4.Write a program to find the sum of first N natural numbers using a loop

```python
x = int(input("Enter the number:"))
num = 0
for i in range(1,x+1):
    num += i
print("add:",num)
```

## 8.Write a function is _even(n) that returns true if the number is even,otherwise false.

```python
def is_even(n):
    if n % 4 == 0:
        return True
    else:
        return False
is_even(8)
```

```python
def max_of_two(a,b):
    return max(a,b)
max_of_two(33,13)
```
